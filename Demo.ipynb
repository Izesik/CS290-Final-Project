```python
import pandas as pd
import duckdb
```

```python
con = duckdb.connect('games.duckdb');
```

```python
franchise_data = con.execute('SELECT name, franchise_duration, franchise_genre FROM franchises').df()
```

```python
franchise_data.sort_values('franchise_duration', ascending=False, inplace=True)
```
